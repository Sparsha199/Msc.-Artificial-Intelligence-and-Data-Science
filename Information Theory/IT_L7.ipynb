**Exercise 3. You are given 16 balls, all of which are equal in weight except for one that is either heavier or lighter. You are also given a bizarre two-pan balance that can report only two outcomes: ‘the two sides balance’ or ‘the two sides do not balance’. Design a strategy to determine which is the odd ball in as few uses of the balance as possible.**


Strategy Overview:
Use the balance scale in a way that each weighing divides the possible outcomes as evenly as possible, we need to narrow down the possibilities with each use of the scale. we need a strategy that efficiently reduces the number of possibilities.

Weighing Strategy:
First weighing: Divide the 16 balls into 3 groups:

Group 1: 5 balls
Group 2: 5 balls
Group 3: 6 balls
Weigh Group 1 (5 balls) against Group 2 (5 balls).

If they balance, the odd ball must be in Group 3 (the 6 balls that were not weighed).
If they do not balance, the odd ball must be in the group that is either heavier or lighter, depending on which side is unbalanced.

Second weighing: Based on the outcome of the first weighing:

Case 1: Group 1 and Group 2 balance (the odd ball is in Group 3)

The odd ball is in Group 3 (6 balls). Now, we split Group 3 into two groups of 3 balls each.
Weigh 3 balls from Group 3 against 3 balls from Group 3.
If they balance, the odd ball is one of the 3 remaining unweighed balls, and you know it is either heavier or lighter.
If they do not balance, the odd ball is in the heavier or lighter group, depending on which side is unbalanced.

Case 2: Group 1 and Group 2 do not balance (the odd ball is in the heavier or lighter group)
Without loss of generality, we assume Group 1 is heavier. Now,we  split Group 1 into two smaller groups (say, 2 balls each) and weigh them.
If the two smaller groups balance, the odd ball must be one of the 3 remaining balls from Group 1 (since we know it is the odd ball).
If they do not balance, we know the odd ball is in the heavier group, and you can further determine whether it is heavier or lighter.

Case 3: We have narrowed down the possibilities to 1 or 2 balls. so use the third weighing to finalize the identification:
Weigh the last 1 or 2 balls to identify the odd ball and determine whether it is heavier or lighter.

Conculsion:
The odd ball can be identified with at most 3 weighings.
By dividing the balls strategically into groups of 5, 5, and 6 in the first weighing and following the above process, we can ensure that each weighing maximizes the information gained, leading to the odd ball efficiently.

**Exercise 6.Let PY = {0.5, 0.5}. Sketch 1/N H δ(Y N ) as a function of δ for N = 1, 2, 3 and 100.**





```python
import numpy as np
import matplotlib.pyplot as plt

P_Y = [0.5, 0.5]
H_PY = -np.sum([p * np.log2(p) for p in P_Y])
delta = np.linspace(0, 2, 500)

N_values = [1, 2, 3, 100]
plt.figure(figsize=(10, 6))


for N in N_values:
    H_delta_N = (H_PY - delta) / N
    plt.plot(delta, H_delta_N, label=f'N={N}')


plt.axhline(0, color='gray', linestyle='--', linewidth=0.8)


plt.xlabel('Delta (δ)')
plt.ylabel('1/N * Hδ(Y^N)')
plt.title('1/N * Hδ(Y^N) as a function of δ')
plt.legend()
plt.grid(True)
plt.show()

```
